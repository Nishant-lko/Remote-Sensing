```python
import numpy as np 
from datasets import load_dataset
```

```python
ds = load_dataset("giswqs/EuroSAT_MS")
```

```python
ds_test = ds['test']
ds_cv = ds['validation']
```

```python
# mean and std from training dataset 
Mean = [0.13543268, 0.11175188, 0.10418837, 0.09467538, 0.11996479, 0.20028183,
 0.23731175, 0.23004931, 0.07332496, 0.0012104,  0.18213873, 0.11190669,
 0.25987505]
Std  = [0.02457886, 0.03333259 ,0.0395221,  0.05942802, 0.05672322, 0.08590868,
 0.10834781 ,0.11147044 ,0.04039885, 0.00047205, 0.10015353, 0.07607304,
 0.12276264]

Mean = np.array(Mean, dtype=np.float32)
Std = np.array(Std, dtype=np.float32)

X_test = [
    (np.array(img, dtype=np.float32) / 10000.0 - Mean[:, None, None])
    / Std[:, None, None]
    for img in ds_test['image']
]

X_cv = [
    (np.array(img, dtype=np.float32) / 10000.0 - Mean[:, None, None])
    / Std[:, None, None]
    for img in ds_cv['image']
]
```

```python
Y_test = [label.split('/')[0] for label in ds_test['filename']]
Y_cv = [label.split('/')[0] for label in ds_cv['filename']]
```

```python
classes = sorted(set(Y_test))
classes
```

```python
import torch 

class_to_idx = {
    name: idx
    for idx, name in enumerate(classes)
}

print(class_to_idx)

Y_cv = torch.tensor(
    [class_to_idx[name] for name in Y_cv],
    dtype=torch.long
)

Y_test = torch.tensor(
    [class_to_idx[name] for name in Y_test],
    dtype=torch.long
)
```

```python
torch.save(X_cv, "X_cv.pt")
torch.save(Y_cv, "Y_cv.pt")

torch.save(X_test,"X_test.pt")
torch.save(Y_test, "Y_test.pt")
```
